```python
import numpy as np
import pandas as pd

# Function to remove duplicates from X and Y arrays
def remove_duplicates_with_labels(X, y, name):
    # Combine features and labels into a DataFrame to handle duplicates row-wise
    # Convert X to a list of tuples to make it hashable for DataFrame deduplication
    # if X is multi-dimensional. Otherwise, DataFrame(X) works directly.
    if X.ndim > 1:
        df_combined = pd.DataFrame(np.hstack((X, y.reshape(-1, 1))))
    else:
        df_combined = pd.DataFrame({'features': X, 'labels': y})

    initial_shape = df_combined.shape[0]
    df_unique = df_combined.drop_duplicates()
    num_removed = initial_shape - df_unique.shape[0]

    if X.ndim > 1:
        new_X = df_unique.iloc[:, :-1].values.astype(X.dtype)
        new_y = df_unique.iloc[:, -1].values.astype(y.dtype)
    else:
        new_X = df_unique['features'].values.astype(X.dtype)
        new_y = df_unique['labels'].values.astype(y.dtype)

    print(f"Removed {num_removed} duplicate rows from {name}_x and {name}_y.")
    print(f"New shape of {name}_x: {new_X.shape}")
    print(f"New shape of {name}_y: {new_y.shape}\n")

    return new_X, new_y

# Remove duplicates for training set
data['train_x'], data['train_y'] = remove_duplicates_with_labels(data['train_x'], data['train_y'], 'train')

# Remove duplicates for validation set
data['val_x'], data['val_y'] = remove_duplicates_with_labels(data['val_x'], data['val_y'], 'val')

# Remove duplicates for test set
data['test_x'], data['test_y'] = remove_duplicates_with_labels(data['test_x'], data['test_y'], 'test')

print("Duplicate rows removed from train, validation, and test sets.")
```

## Perform Feature Engineering

### Subtask:
Create two new numerical features for each sequence in `train_x`, `val_x`, and `test_x`: effective length (count of non-padding characters where '0' is the padding token) and number of unique non-padding characters. Append these features to the respective feature arrays and then re-scale all augmented feature arrays using `MinMaxScaler`.


**Reasoning**:
To perform feature engineering, I will define a function to calculate 'effective length' and 'number of unique characters' for each sequence in the feature arrays, then apply this function to the training, validation, and test sets. After generating these new features, I will append them to the existing feature arrays.



```python
import numpy as np
from sklearn.preprocessing import MinMaxScaler

def engineer_features(X):
    effective_length = []
    unique_chars = []
    for row in X:
        non_padding_elements = row[row != 0] # '0' is the padding token
        effective_length.append(len(non_padding_elements))
        unique_chars.append(len(np.unique(non_padding_elements)))
    return np.array([effective_length, unique_chars]).T

# Generate new features for each set
new_train_features = engineer_features(data['train_x'])
new_val_features = engineer_features(data['val_x'])
new_test_features = engineer_features(data['test_x'])

# Horizontally stack the new features with the existing feature arrays
data['train_x'] = np.hstack((data['train_x'], new_train_features))
data['val_x'] = np.hstack((data['val_x'], new_val_features))
data['test_x'] = np.hstack((data['test_x'], new_test_features))

print("New features (effective length and unique characters) have been generated and appended.")
print(f"New shape of data['train_x']: {data['train_x'].shape}")
print(f"New shape of data['val_x']: {data['val_x'].shape}")
print(f"New shape of data['test_x']: {data['test_x'].shape}")
```

**Reasoning**:
Following the instructions, I will re-instantiate the `MinMaxScaler`, fit it on the augmented `data['train_x']`, and then transform all augmented feature arrays. Finally, I will print the new shapes and the first 5 rows of the scaled `data['train_x']` to verify the transformation and the presence of the new features.



```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np

# Re-instantiate MinMaxScaler
scaler = MinMaxScaler()

# Fit the scaler on the augmented data['train_x']
scaler.fit(data['train_x'])

# Transform all augmented feature arrays
data['train_x'] = scaler.transform(data['train_x'])
data['val_x'] = scaler.transform(data['val_x'])
data['test_x'] = scaler.transform(data['test_x'])

print("Augmented numerical features in train_x, val_x, and test_x have been re-scaled to a [0, 1] range.")

# Print the new shapes of the scaled data arrays
print(f"\nNew shape of scaled data['train_x']: {data['train_x'].shape}")
print(f"New shape of scaled data['val_x']: {data['val_x'].shape}")
print(f"New shape of scaled data['test_x']: {data['test_x'].shape}")

# Print the first 5 rows of the scaled data['train_x']
print("\nFirst 5 rows of scaled data['train_x']:")
print(data['train_x'][:5])
```

## Prepare Data for Modeling

### Subtask:
Ensure the training and testing data are correctly formatted and valid for scikit-learn modeling.


**Reasoning**:
I will check the shapes, data types, and check for infinite or NaN values in the training and testing sets to ensure compatibility with scikit-learn.



```python
import numpy as np

def validate_data(X, y, name):
    print(f"--- Validating {name} data ---")
    print(f"Shape: X={X.shape}, y={y.shape}")
    print(f"Dtypes: X={X.dtype}, y={y.dtype}")

    # Check for Inf or NaN
    is_finite = np.all(np.isfinite(X))
    has_nan = np.any(np.isnan(X))
    print(f"All values finite: {is_finite}")
    print(f"Contains NaN: {has_nan}")

    # Ensure labels are numeric integers for classification
    if not np.issubdtype(y.dtype, np.integer):
        print(f"Casting {name}_y to int32")
        y = y.astype(np.int32)

    return y

# Validate and potentially cast labels
data['train_y'] = validate_data(data['train_x'], data['train_y'], 'train')
data['test_y'] = validate_data(data['test_x'], data['test_y'], 'test')

print("\nData preparation for scikit-learn is complete.")
```
