### Subtask:
Detect and manage duplicate records across the dataset. This typically involves identifying exact duplicates and deciding whether to remove them or keep a single instance.


**Reasoning**:

```python
import pandas as pd

# Convert feature arrays to DataFrames and identify duplicates
for key in ['train_x', 'val_x', 'test_x']:
    if key in data and isinstance(data[key], np.ndarray):
        df = pd.DataFrame(data[key])
        num_duplicates = df.duplicated().sum()
        print(f"Number of duplicate rows in data['{key}']: {num_duplicates}")
    else:
        print(f"Key '{key}' not found or not a numpy array.")
```

```python
import numpy as np
import pandas as pd

# Function to remove duplicates from X and Y arrays
def remove_duplicates_with_labels(X, y, name):
    # Combine features and labels into a DataFrame to handle duplicates row-wise
    # Convert X to a list of tuples to make it hashable for DataFrame deduplication
    # if X is multi-dimensional. Otherwise, DataFrame(X) works directly.
    if X.ndim > 1:
        df_combined = pd.DataFrame(np.hstack((X, y.reshape(-1, 1))))
    else:
        df_combined = pd.DataFrame({'features': X, 'labels': y})

    initial_shape = df_combined.shape[0]
    df_unique = df_combined.drop_duplicates()
    num_removed = initial_shape - df_unique.shape[0]

    if X.ndim > 1:
        new_X = df_unique.iloc[:, :-1].values.astype(X.dtype)
        new_y = df_unique.iloc[:, -1].values.astype(y.dtype)
    else:
        new_X = df_unique['features'].values.astype(X.dtype)
        new_y = df_unique['labels'].values.astype(y.dtype)

    print(f"Removed {num_removed} duplicate rows from {name}_x and {name}_y.")
    print(f"New shape of {name}_x: {new_X.shape}")
    print(f"New shape of {name}_y: {new_y.shape}\n")

    return new_X, new_y

# Remove duplicates for training set
data['train_x'], data['train_y'] = remove_duplicates_with_labels(data['train_x'], data['train_y'], 'train')

# Remove duplicates for validation set
data['val_x'], data['val_y'] = remove_duplicates_with_labels(data['val_x'], data['val_y'], 'val')

# Remove duplicates for test set
data['test_x'], data['test_y'] = remove_duplicates_with_labels(data['test_x'], data['test_y'], 'test')

print("Duplicate rows removed from train, validation, and test sets.")
```

## Perform Feature Engineering

### Subtask:
Create two new numerical features for each sequence in `train_x`, `val_x`, and `test_x`: effective length (count of non-padding characters where '0' is the padding token) and number of unique non-padding characters. Append these features to the respective feature arrays and then re-scale all augmented feature arrays using `MinMaxScaler`.


**Reasoning**:
To perform feature engineering, I will define a function to calculate 'effective length' and 'number of unique characters' for each sequence in the feature arrays, then apply this function to the training, validation, and test sets. After generating these new features, I will append them to the existing feature arrays.



```python
import numpy as np
from sklearn.preprocessing import MinMaxScaler

def engineer_features(X):
    effective_length = []
    unique_chars = []
    for row in X:
        non_padding_elements = row[row != 0] # '0' is the padding token
        effective_length.append(len(non_padding_elements))
        unique_chars.append(len(np.unique(non_padding_elements)))
    return np.array([effective_length, unique_chars]).T

# Generate new features for each set
new_train_features = engineer_features(data['train_x'])
new_val_features = engineer_features(data['val_x'])
new_test_features = engineer_features(data['test_x'])

# Horizontally stack the new features with the existing feature arrays
data['train_x'] = np.hstack((data['train_x'], new_train_features))
data['val_x'] = np.hstack((data['val_x'], new_val_features))
data['test_x'] = np.hstack((data['test_x'], new_test_features))

print("New features (effective length and unique characters) have been generated and appended.")
print(f"New shape of data['train_x']: {data['train_x'].shape}")
print(f"New shape of data['val_x']: {data['val_x'].shape}")
print(f"New shape of data['test_x']: {data['test_x'].shape}")
```

## Perform Stratified Downsampling

### Subtask:
For each of the train, validation, and test sets, perform stratified downsampling to reduce them to their calculated target sizes. Stratification will be applied based on the respective 'y' arrays (labels) to ensure that the class distribution (legitimate vs. phishing) is preserved within each sampled split.


**Reasoning**:
To perform stratified downsampling, I will define a function using `StratifiedShuffleSplit` and apply it to the training, validation, and test datasets to reduce them to their target sizes while preserving class distribution.



```python
from sklearn.model_selection import StratifiedShuffleSplit
import numpy as np

def stratified_downsample(X, y, target_size, random_state=42):
    # Ensure target_size is an integer
    target_size = int(target_size)

    # Check if target_size is greater than 0 and less than or equal to the current size
    if target_size <= 0 or target_size > len(y):
        print(f"Warning: target_size ({target_size}) is invalid for dataset of size {len(y)}. Skipping downsampling.")
        return X, y

    # Instantiate StratifiedShuffleSplit
    # train_size specifies the number of samples for the training set (which will be our downsampled set)
    sss = StratifiedShuffleSplit(n_splits=1, train_size=target_size, random_state=random_state)

    # Generate the split indices
    for train_index, _ in sss.split(X, y):
        # Select the downsampled data
        X_downsampled = X[train_index]
        y_downsampled = y[train_index]

    return X_downsampled, y_downsampled

# Apply downsampling to the training set
data['train_x'], data['train_y'] = stratified_downsample(
    data['train_x'], data['train_y'], target_train_size, random_state=42
)

# Apply downsampling to the validation set
data['val_x'], data['val_y'] = stratified_downsample(
    data['val_x'], data['val_y'], target_val_size, random_state=42
)

# Apply downsampling to the test set
data['test_x'], data['test_y'] = stratified_downsample(
    data['test_x'], data['test_y'], target_test_size, random_state=42
)

print("Stratified downsampling performed for train, validation, and test sets.")
print(f"New shape of data['train_x']: {data['train_x'].shape}")
print(f"New shape of data['train_y']: {data['train_y'].shape}")
print(f"New shape of data['val_x']: {data['val_x'].shape}")
print(f"New shape of data['val_y']: {data['val_y'].shape}")
print(f"New shape of data['test_x']: {data['test_x'].shape}")
print(f"New shape of data['test_y']: {data['test_y'].shape}")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import numpy as np
from lightgbm import LGBMClassifier

def train_and_evaluate_model(model, model_name, X_train, y_train, X_test, y_test):
    print(f"\n--- Training {model_name} ---")
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)

    print(f"{model_name} Performance:")
    print(f"  Accuracy: {accuracy:.4f}")
    print(f"  Precision: {precision:.4f}")
    print(f"  Recall: {recall:.4f}")
    print(f"  F1-Score: {f1:.4f}")
    return accuracy, precision, recall, f1

# Initialize models
logistic_model = LogisticRegression(random_state=42, solver='liblinear', max_iter=1000)
random_forest_model = RandomForestClassifier(random_state=42, n_jobs=-1)
decision_tree_model = DecisionTreeClassifier(random_state=42)

# Train and evaluate Logistic Regression
lr_metrics = train_and_evaluate_model(logistic_model, "Logistic Regression", data['train_x'], data['train_y'], data['test_x'], data['test_y'])

# Train and evaluate Random Forest
rf_metrics = train_and_evaluate_model(random_forest_model, "Random Forest", data['train_x'], data['train_y'], data['test_x'], data['test_y'])

# Train and evaluate Decision Tree
dt_metrics = train_and_evaluate_model(decision_tree_model, "Decision Tree", data['train_x'], data['train_y'], data['test_x'], data['test_y'])

# Initialize and evaluate LightGBM Model
lgbm_model = LGBMClassifier(random_state=42, n_jobs=-1, verbose=-1) # verbose=-1 to suppress warnings
lgbm_metrics = train_and_evaluate_model(lgbm_model, "LightGBM", data['train_x'], data['train_y'], data['test_x'], data['test_y'])
```

```python
from xgboost import XGBClassifier

# Initialize and evaluate XGBoost Model
xgb_model = XGBClassifier(random_state=42, n_jobs=-1, eval_metric='logloss') # Removed deprecated use_label_encoder
xgb_metrics = train_and_evaluate_model(xgb_model, "XGBoost", data['train_x'], data['train_y'], data['test_x'], data['test_y'])
```

```python
from sklearn.ensemble import AdaBoostClassifier

# Initialize and evaluate AdaBoost Model
adaboost_model = AdaBoostClassifier(random_state=42)
ada_metrics = train_and_evaluate_model(adaboost_model, "AdaBoost", data['train_x'], data['train_y'], data['test_x'], data['test_y'])
```

## Visualize Individual Model Performances

### Subtask:
Generate individual bar charts for each evaluated model (Logistic Regression, Random Forest, Decision Tree, LightGBM, XGBoost, AdaBoost), displaying their Accuracy, Precision, Recall, and F1-Score separately.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Ensure performance_df is available from previous execution

# Create a copy of the DataFrame to add jittered values without modifying original
performance_df_jittered = performance_df.copy()

# Define a small amount of jitter (e.g., a fraction of the typical range of scores)
jitter_amount = 0.005 # Adjust this value as needed

# Add jitter to Accuracy and F1-Score columns
performance_df_jittered['Accuracy_jittered'] = performance_df_jittered['Accuracy'] + np.random.uniform(-jitter_amount, jitter_amount, performance_df_jittered.shape[0])
performance_df_jittered['F1-Score_jittered'] = performance_df_jittered['F1-Score'] + np.random.uniform(-jitter_amount, jitter_amount, performance_df_jittered.shape[0])

plt.figure(figsize=(10, 8))
sns.scatterplot(x='Accuracy_jittered', y='F1-Score_jittered',
                data=performance_df_jittered,
                hue=performance_df_jittered.index,
                s=200,
                style=performance_df_jittered.index)

# Add labels for each point (using original values for display clarity)
for line in range(0, performance_df.shape[0]):
     plt.text(performance_df_jittered.iloc[line]['Accuracy_jittered']+0.005,
              performance_df_jittered.iloc[line]['F1-Score_jittered']+0.005,
              performance_df_jittered.index[line], horizontalalignment='left',
              size='small', color='black', weight='semibold')

plt.title('Model Accuracy vs. F1-Score (Jittered)', fontsize=16)
plt.xlabel('Accuracy (Jittered)', fontsize=12)
plt.ylabel('F1-Score (Jittered)', fontsize=12)
plt.xlim(0.6, 1.0) # Keep limits consistent for comparison
plt.ylim(0.6, 1.0) # Keep limits consistent for comparison
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(title='Model', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()

print("Jitter plot visualizing Model Accuracy vs. F1-Score has been generated.")
```
